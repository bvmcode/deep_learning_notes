# Neural Networks

* Deep learning is a subset of machine learning and artificial intelligence (AI) that uses multi-layered artificial neural networks (ANNs) to automatically learn complex patterns and representations from raw, unstructured data like images, text, and audio.
* Unlike traditional machine learning, which requires humans to manually select and code features, deep learning models discover abstract features on their own by passing data through layers of interconnected processing nodes.
* Consider a simple example like predicting the price of a home from its size. The neural network has one node. The input X is to the node is the house size and output Y is price. The node itself computes a function like $wx+b$ and outputs the estimated price Y.

* Now let's consider more features like number of bedrooms, zip code and wealth. Each of these would feed into multiple nodes. The nodes perform a calculation and feed into an output node. So X is in the input data, the hidden layer is the layer of nodes that take in the input. We do not need to explicitly provide the intermediate concepts such as family size, walkability and school quality. The network learns internal functions that are useful for mapping X to Y. This is one of the major strengths of neural networks. Given enough appropriate training data, neural networks can become very effective at learning complicated mappings between inputs and outputs.

* Different types of neural networks

| Neural Network Type | What It’s Good At | Common Use Cases |
|---|---|---|
| **Feedforward Neural Network (FNN / MLP)** | Learning relationships in structured/tabular data | Classification, regression, credit scoring, basic prediction |
| **Convolutional Neural Network (CNN)** | Recognizing spatial patterns, especially in images | Image classification, object detection, medical imaging, facial recognition |
| **Recurrent Neural Network (RNN)** | Processing sequential data where order matters | Time-series prediction, text processing, speech recognition |
| **LSTM (Long Short-Term Memory)** | Handling longer-term dependencies in sequences; improved form of RNN | Time-series forecasting, speech, language modeling, anomaly detection |
| **GRU (Gated Recurrent Unit)** | Similar to LSTM but simpler and often faster | Time-series forecasting, sequence classification, speech/NLP |
| **Transformer** | Modeling relationships across long sequences efficiently | LLMs, translation, summarization, code generation, increasingly vision/audio |
| **Autoencoder (AE)** | Learning compressed representations of data | Dimensionality reduction, anomaly detection, denoising, feature learning |
| **Variational Autoencoder (VAE)** | Learning a probabilistic representation that can generate new samples | Image generation, synthetic data, representation learning |
| **Generative Adversarial Network (GAN)** | Generating realistic synthetic data using competing generator/discriminator networks | Image generation, image enhancement, synthetic data, style transfer |
| **Graph Neural Network (GNN)** | Learning from data represented as nodes and relationships/edges | Social networks, fraud detection, recommendation systems, molecular modeling |
| **Diffusion Model** | Generating complex data by learning to reverse a gradual noise process | Image generation/editing, video generation, audio generation |
| **Encoder–Decoder / Seq2Seq** | Transforming one sequence or representation into another | Translation, summarization, speech-to-text, image captioning |

* Thanks to DL, computers are much better at unstructured data.

| Data Type | Definition | Examples |
|---|---|---|
| **Structured Data** | Data organized according to a predefined structure or schema, typically in rows and columns with clearly defined fields and data types. | SQL databases, Excel tables, customer records, financial transactions, sales data, sensor readings |
| **Unstructured Data** | Data that does not follow a predefined tabular structure or schema. Its information is often contained within text, images, audio, video, or other complex formats. | Emails, PDFs, Word documents, images, medical notes, audio recordings, videos, social media posts |

* Deep learning's rise has been driven primarily by three factors: more data, more computing power, and better algorithms. Digitization, mobile devices, cameras, and IoT sensors have dramatically increased the amount of available training data.
* Large neural networks benefit more from large datasets. Traditional ML algorithms may eventually plateau as data increases, while sufficiently large neural networks can continue improving with additional data.
* Scale matters on two dimensions: the size of the neural network (parameters/connections) and the amount of labeled training data. With small datasets, traditional algorithms can still perform competitively, especially with good feature engineering.
* Algorithmic improvements have made training much faster. For example, replacing sigmoid activation functions with ReLU helps reduce problems with near-zero gradients and speeds up gradient descent.
* Faster computation accelerates experimentation. Shorter training times allow researchers to repeatedly test ideas, evaluate results, modify models, and discover better architectures much faster. 
* These drivers are still improving, with growing datasets, specialized hardware such as GPUs, faster networking, and continued algorithmic innovation suggesting further progress in deep learning.
