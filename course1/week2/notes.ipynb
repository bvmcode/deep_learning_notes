### Logistic Regression

* Logistic regression is a statistical and machine learning method used to predict a categorical outcome. It calculates the probability that a data point belongs to a specific class, usually resulting in a value between 0 and 1. 
* Here $\hat{y}$ is the probability an input belongs to class 1.

$$\hat{y} = P(y=1 | x)$$
$$
x \in \mathbb{R}^{n_x}
$$

* The model has two parameters where $w$ is a vector of weights with the same number of dimensions as the input feature and $b$ is a scalar bias and intercept.

$$
w \in \mathbb{R}^{n_x}
$$
$$b \in \mathbb{R}$$

* The simple linear model $z = w^T x + b$ doesn't work well as a probability because its output can be less than 0 or greater than 1. So we must introduce a non-linear activation that outputs values between 0 and 1, which is the sigmoid.

$$\hat{y} = \sigma(z) = \frac{1}{1+e^z}$$

* The sigmoid maps any real number into the range 0 to 1:
    * If `z>>0` then $\hat{y} \to 1$.
    * If `z=0`then $\hat{y} =.5$.
    * If `z<<00` then $\hat{y} \to 0$.

<img src="./images/2.png" height=500 width=800/>

* The superscript `i` means a single example

$$z^{(i)} = w^T x^{(i)} + b$$
$$\hat{y} = \sigma(w^T x^{(i)} + b) = \frac{1}{1+e^{z^{(i)}}} $$


### Logistic Regression Cost Function

* A loss function calculates the error for a single data point, while a cost function measures the average or total error across the entire training dataset.
* We need a loss function to see how good $\hat{y}$ is compared to $y$. From that We need to use a cost function that is convex so we can find the minimum with gradient descent.
* Logistic regression uses **log loss**, also called binary cross-entropy loss. This loss strongly rewards confident correct predictions and strongly penalizes confident incorrect ones.

$$
L(\hat{y},y)
=
-\left[
y\log(\hat{y})
+
(1-y)\log(1-\hat{y})
\right]
$$

$$
L\left(\hat{y}^{(i)},y^{(i)}\right)
=
-\left[
y^{(i)}\log\left(\hat{y}^{(i)}\right)
+
\left(1-y^{(i)}\right)
\log\left(1-\hat{y}^{(i)}\right)
\right]
$$

* Case 1 - The true label is $y=1$. So when the actual label is 1, training encourages the model to predict a probability close to 1.

$$
L(\hat{y},1)
=
-\log(\hat{y})
$$

$$
\hat{y} \rightarrow 1
$$

* Case 2 - The true label is $y=0$. So when the actual label is 0, training encourages the model to predict a probability close to 0.

$$
L(\hat{y},0)
=
-\log(1-\hat{y})
$$

$$
\hat{y} \rightarrow 0
$$

* The cost function measures performance over the entire training set by averaging the individual losses:

$$
J(w,b)
=
\frac{1}{m}
\sum_{i=1}^{m}
L\left(\hat{y}^{(i)},y^{(i)}\right)
$$

$$\text{Substituting the logistic loss gives:}$$

$$
J(w,b)
=
-\frac{1}{m}
\sum_{i=1}^{m}
\left[
y^{(i)}
\log\left(\hat{y}^{(i)}\right)
+
\left(1-y^{(i)}\right)
\log\left(1-\hat{y}^{(i)}\right)
\right]
$$

* Training logistic regression means finding `w` and `b` that minimize the cost:

$$
\min_{w,b} J(w,b)
$$

* In short: **Linear model → Sigmoid → Prediction → Loss per example → Average loss (cost) → Minimize cost to learn $w,b$.**

* This logistic regression model can also be interpreted as a very small neural network, which provides a bridge from logistic regression to more general neural-network architectures.

### Gradient Descent

* Gradient descent is an **optimization algorithm** used to find the values of model parameters that minimize a cost function.

* For logistic regression, the model parameters are:
  * $w$ — the weight vector
  * $b$ — the bias term

* The goal of training is to find values of $w$ and $b$ that make the cost function $J(w,b)$ as small as possible:

$$
\min_{w,b} J(w,b)
$$


* The loss function measures how well the model performs on a **single training example**. The cost function measures performance over the **entire training set** by averaging the individual losses:

$$
J(w,b)
=
\frac{1}{m}
\sum_{i=1}^{m}
L\left(\hat{y}^{(i)},y^{(i)}\right)
$$

* For logistic regression, the individual loss is the binary cross-entropy loss:

$$
L\left(\hat{y}^{(i)},y^{(i)}\right)
=
-
\left[
y^{(i)}\log\left(\hat{y}^{(i)}\right)
+
\left(1-y^{(i)}\right)
\log\left(1-\hat{y}^{(i)}\right)
\right]
$$

* Therefore, the full logistic-regression cost function is:

$$
J(w,b)
=
-\frac{1}{m}
\sum_{i=1}^{m}
\left[
y^{(i)}\log\left(\hat{y}^{(i)}\right)
+
\left(1-y^{(i)}\right)
\log\left(1-\hat{y}^{(i)}\right)
\right]
$$

* Training logistic regression means finding $w$ and $b$ that minimize:

$$
J(w,b)
$$

* The logistic-regression cost function is **convex**. A convex cost function can be visualized as a large bowl with a single global minimum.

* Because the function is convex, gradient descent should converge to the same global minimum, or approximately the same minimum, regardless of a reasonable initialization.

* Start with initial values for $w$ and $b$. For logistic regression, the parameters are commonly initialized to zero:

$$
w = 0
$$

$$
b = 0
$$

* Gradient descent then repeatedly takes a step in the **steepest downhill direction** of the cost function. Each step should reduce:

$$
J(w,b)
$$

* The process continues until the parameters converge near the minimum of the cost function. To understand the intuition, first imagine that the cost function depends only on $w$:

$$
J(w)
$$

* Gradient descent repeatedly performs the following update:

$$
w
:=
w
-
\alpha
\frac{dJ(w)}{dw}
$$

$$\text{where:}$$

* $\alpha$ is the **learning rate**. $\frac{dJ(w)}{dw}$ is the derivative, or slope, of the cost function at the current value of $w$. The learning rate $\alpha$ controls the size of each gradient-descent step.

$$
\alpha = \text{step size}
$$

* A larger $\alpha$ produces larger parameter updates. A smaller $\alpha$ produces smaller parameter updates.

* Gradient descent repeatedly uses the learning rate to determine how far to move in the downhill direction. The derivative tells us the **slope** of the cost function at the current parameter value.

$$
\frac{dJ(w)}{dw}
$$

* The sign of the derivative determines which direction $w$ should move.

* Case 1 — Positive derivative

$$\text{If:}$$

$$
\frac{dJ(w)}{dw} > 0
$$

$$\text{then:}$$

$$
w
:=
w
-
\alpha
\left(\text{positive number}\right)
$$

$$\text{so:}$$

$$
w \downarrow
$$

$$\text{Therefore, $w$ moves to the left toward the minimum.}$$


* Case 2 — Negative derivative

$$\text{If:}$$

$$
\frac{dJ(w)}{dw} < 0
$$

$$\text{then:}$$

$$
w
:=
w
-
\alpha
\left(\text{negative number}\right)
$$

$$\text{which means:}$$

$$
w
:=
w
+
\text{positive amount}
$$

$$\text{so:}$$

$$
w \uparrow
$$

$$\text{Therefore, $w$ moves to the right toward the minimum.}$$


* Logistic regression has both $w$ and $b$, so both parameters must be updated. These updates are repeated until the algorithm converges.

$$\text{Update $w$}$$

$$
w
:=
w
-
\alpha
\frac{\partial J(w,b)}{\partial w}
$$

$$\text{Update $b$}$$

$$
b
:=
b
-
\alpha
\frac{\partial J(w,b)}{\partial b}
$$


* Partial Derivatives - When a function has only one variable, we normally use an ordinary derivative:

$$
\frac{dJ(w)}{dw}
$$

* When a function depends on multiple variables, such as:

$$
J(w,b)
$$

$$\text{we use a partial derivative:}$$

$$
\frac{\partial J(w,b)}{\partial w}
$$

$$\text{and:}$$

$$
\frac{\partial J(w,b)}{\partial b}
$$

* A partial derivative measures how the cost changes when changing **one parameter while considering the others fixed**.


* When gradient descent is implemented in code, the derivatives are commonly represented using:

$$
dw
=
\frac{\partial J(w,b)}{\partial w}
$$

$$
db
=
\frac{\partial J(w,b)}{\partial b}
$$

$$\text{The parameter updates therefore become:}$$

$$
w := w - \alpha dw
$$

$$
b := b - \alpha db
$$

* Overall Training Process - Logistic regression first produces a prediction:

$$
z = w^Tx + b
$$

$$
\hat{y} = \sigma(z)
$$

* The prediction is compared with the true label using the loss function:

$$
L(\hat{y},y)
$$

* Losses across all training examples are averaged to obtain:

$$
J(w,b)
$$

* Gradient descent computes the slope of the cost function. The parameters are updated:

$$
w := w - \alpha dw
$$

$$
b := b - \alpha db
$$

* The process repeats until the cost reaches or approaches its minimum.

* Summary - **Input → Linear Model → Sigmoid → Prediction → Loss → Cost → Compute Gradients → Update $(w,b)$ → Repeat**, or:

$$
x
\rightarrow
z=w^Tx+b
\rightarrow
\hat{y}=\sigma(z)
\rightarrow
L(\hat{y},y)
\rightarrow
J(w,b)
\rightarrow
(dw,db)
\rightarrow
(w,b)\text{ update}
$$

* The key intuition is simple - **Gradient descent looks at the slope of the cost function and repeatedly moves the model parameters downhill until it reaches the minimum.**

### Derivatives: Intuition and Meaning

- You **do not need a deep understanding of calculus** to successfully apply neural networks and deep learning.
- The main goal is to develop an **intuitive understanding of derivatives**.
- Later, **forward functions** and **backward functions** will encapsulate much of the calculus needed for deep learning.
- A simple way to think about a **derivative** is as the **slope of a function**.

- Consider the function:

$$
f(a)=3a
$$

- This is a straight line, so we can understand its derivative by looking at how the output changes when the input changes by a tiny amount.

$$
\text{At:}\qquad a=2
$$

$$
f(a)=3(2)=6
$$

- Give $a$ a tiny nudge to the right:

$$
a=2.001
$$

- The new value of the function is:

$$
f(a)=3(2.001)=6.003
$$

- Therefore, the change in $a$ is:

$$
\Delta a=0.001
$$

- And the corresponding change in $f(a)$ is:

$$
\Delta f=0.003
$$

- The **slope**, or **derivative**, is the height divided by the width:

$$
\text{slope}
=
\frac{\Delta f}{\Delta a}
=
\frac{0.003}{0.001}
=
3
$$

- This means that when $a$ is nudged slightly to the right, $f(a)$ increases by **3 times as much**.

$$
\text{Derivative}=\text{Slope}
$$

- Now consider another point on the same function:

$$
\text{At:}\qquad a=5
$$

$$
f(a)=3(5)=15
$$

- Give $a$ another tiny nudge:

$$
a=5.001
$$

$$
f(a)=3(5.001)=15.003
$$

- Once again:

$$
\Delta a=0.001
$$

$$
\Delta f=0.003
$$

- So the slope is:

$$
\frac{\Delta f}{\Delta a}
=
\frac{0.003}{0.001}
=
3
$$

- Therefore, the derivative at $a=5$ is also:

$$
3
$$

- The derivative of $f(a)$ with respect to $a$ is written:

$$
\frac{df(a)}{da}
$$

- An equivalent notation is:

$$
\frac{d}{da}f(a)
$$

- For this function:

$$
f(a)=3a
$$

$$\text{we have:}$$

$$
\frac{df(a)}{da}=3
$$

- Intuitively, this means that if $a$ is increased by a tiny amount, $f(a)$ increases by approximately **3 times that amount**:

$$
\Delta f\approx3\Delta a
$$

- In the examples above, a small change of:

$$
\Delta a=0.001
$$

$$\text{produced:}$$

$$
\Delta f=3(0.001)=0.003
$$

- The formal mathematical definition considers an **infinitesimally small** change rather than specifically using $0.001$:

$$
\frac{df(a)}{da}
=
\lim_{\Delta a\to0}
\frac{f(a+\Delta a)-f(a)}{\Delta a}
$$

- Using $0.001$ is simply a convenient way to build intuition for what happens when the input changes by a very small amount.

- One important property of:

$$
f(a)=3a
$$

$$\text{is that its slope is the same everywhere.}$$

$$
\text{At:}\qquad a=2
$$

$$
\frac{df(a)}{da}=3
$$

$$
\text{At:}\qquad a=5
$$

$$
\frac{df(a)}{da}=3
$$

$$\text{In fact, for any value of $a$:}$$

$$
\frac{df(a)}{da}=3
$$

$$\text{Geometrically, wherever a small slope triangle is drawn on the line, its height-to-width ratio is always:}$$

$$
\frac{\text{height}}{\text{width}}=3
$$

$$\text{Key intuition:** the derivative tells us how much the output of a function changes when its input changes by a tiny amount.}$$

$$
f(a)=3a
\qquad\Longrightarrow\qquad
\frac{df(a)}{da}=3
$$

$$\text{So, for a tiny change in $a$:}$$

$$
\text{Change in }f(a)
\approx
3\times\text{Change in }a
$$

### Derivatives of non-linear functions

- For a **nonlinear function**, the slope can be different at different points on the curve.
- Consider the function:

$$
f(a)=a^2
$$

- Unlike the straight-line function $f(a)=3a$, the slope of $f(a)=a^2$ changes depending on the value of $a$.

$$
\text{At:}\qquad a=2
$$

$$
f(a)=2^2=4
$$

- Give $a$ a tiny nudge to the right:

$$
a=2.001
$$

$$\text{Then:}$$

$$
f(a)=(2.001)^2=4.004001\approx4.004
$$

$$\text{Therefore:}$$

$$
\Delta a=0.001
$$

$$
\Delta f\approx0.004
$$

- So $f(a)$ increases approximately **4 times as much** as $a$:

$$
\frac{\Delta f}{\Delta a}
\approx
\frac{0.004}{0.001}
=
4
$$

- In calculus terminology, the **slope**, or **derivative**, at $a=2$ is:

$$
\text{At:}\qquad a=2
$$

$$
\frac{d}{da}f(a)=4
$$

- Now consider a different point on the same curve:

$$
\text{At:}\qquad a=5
$$

$$
f(a)=5^2=25
$$

- Give $a$ the same tiny nudge:

$$
a=5.001
$$

$$\text{Then:}$$

$$
f(a)=(5.001)^2\approx25.010
$$

$$\text{Therefore:}$$

$$
\Delta a=0.001
$$

$$
\Delta f\approx0.010
$$

- Now $f(a)$ increases approximately **10 times as much** as $a$:

$$
\frac{\Delta f}{\Delta a}
\approx
\frac{0.010}{0.001}
=
10
$$

$$\text{So:}$$

$$
\text{At:}\qquad a=5
$$

$$
\frac{d}{da}f(a)=10
$$

- This illustrates an important difference between straight lines and curves:

$$
\text{At }a=2:\qquad \text{slope}=4
$$

$$
\text{At }a=5:\qquad \text{slope}=10
$$

- Geometrically, the small slope triangles have different **height-to-width ratios** at different locations on the curve.


- Derivative of $a^2$ = A standard calculus formula tells us:

$$
f(a)=a^2
$$

$$
\frac{d}{da}a^2=2a
$$

- This formula gives the slope at **any value of $a$**.

$$
\text{At:}\qquad a=2
$$

$$
\frac{d}{da}a^2=2(2)=4
$$

$$
\text{At:}\qquad a=5
$$

$$
\frac{d}{da}a^2=2(5)=10
$$

- This agrees with the small-nudge examples above. More generally, if $a$ changes by a tiny amount $\Delta a$, the corresponding change in the function is approximately:

$$
\Delta f
\approx
\frac{df}{da}\Delta a
$$

- For $f(a)=a^2$:

$$
\Delta f\approx2a\,\Delta a
$$

$$\text{So when:}$$

$$
a=2,\qquad \Delta a=0.001
$$

$$\text{we predict:}$$

$$
\Delta f
\approx
2(2)(0.001)
=
0.004
$$

- Why the Change Is Only Approximate. When $a$ changes from $2$ to $2.001$, the exact function value is:

$$
(2.001)^2=4.004001
$$

- The derivative predicts approximately:

$$
4+0.004=4.004
$$

- The tiny difference comes from the fact that $0.001$ is small, but it is **not infinitesimally small**. In fact:

$$
(a+\Delta a)^2
=
a^2+2a\Delta a+(\Delta a)^2
$$

- The derivative captures the main change:

$$
2a\Delta a
$$

$$\text{while the additional term is:}$$

$$
(\Delta a)^2
$$

- As the nudge becomes infinitesimally small:

$$
\Delta a\to0
$$

- The extra term becomes negligible, and the derivative gives the instantaneous rate of change.

$$
df
=
\frac{df}{da}\,da
$$

* Example: $f(a)=a^3$. Consider another nonlinear function:

$$
f(a)=a^3
$$

$$\text{Its derivative is:}$$

$$
\frac{d}{da}a^3=3a^2
$$

$$
\text{At:}\qquad a=2
$$

$$
f(a)=2^3=8
$$

- Give $a$ a tiny nudge:

$$
a=2.001
$$

$$\text{Then:}$$

$$
f(a)=(2.001)^3\approx8.012
$$

- The derivative at $a=2$ is:

$$
\frac{d}{da}a^3
=
3(2)^2
=
12
$$

- Therefore, a change of $\Delta a=0.001$ should produce approximately:

$$
\Delta f
\approx
12(0.001)
=
0.012
$$

$$\text{And indeed:}$$

$$
8\longrightarrow8.012
$$

- So the output changes approximately **12 times as much** as the input near $a=2$.


* Example: $f(a)=\ln(a)$. Now consider the natural logarithm:

$$
f(a)=\ln(a)
$$

$$\text{Its derivative is:}$$

$$
\frac{d}{da}\ln(a)=\frac{1}{a}
$$

$$
\text{At:}\qquad a=2
$$

- The function value is approximately:

$$
f(2)=\ln(2)\approx0.69315
$$

- Increase $a$ slightly:

$$
a=2.001
$$

$$\text{Then:}$$

$$
f(2.001)=\ln(2.001)\approx0.69365
$$

$$\text{So:}$$

$$
\Delta a=0.001
$$

$$
\Delta f\approx0.0005
$$

- The derivative predicts exactly this approximate behavior:

$$
\text{At:}\qquad a=2
$$

$$
\frac{d}{da}\ln(a)
=
\frac{1}{2}
$$

$$\text{Therefore:}$$

$$
\Delta f
\approx
\frac{1}{2}\Delta a
$$

$$
\Delta f
\approx
\frac{1}{2}(0.001)
=
0.0005
$$

- In other words, near $a=2$, the output changes by approximately **half as much** as the input.


* Comparing the Examples. Different functions have different derivative formulas:

$$
f(a)=3a
\qquad\Longrightarrow\qquad
\frac{df}{da}=3
$$

$$
f(a)=a^2
\qquad\Longrightarrow\qquad
\frac{df}{da}=2a
$$

$$
f(a)=a^3
\qquad\Longrightarrow\qquad
\frac{df}{da}=3a^2
$$

$$
f(a)=\ln(a)
\qquad\Longrightarrow\qquad
\frac{df}{da}=\frac{1}{a}
$$

- For the straight-line function:

$$
f(a)=3a
$$

$$\text{the derivative is the same everywhere:}$$

$$
\frac{df}{da}=3
$$

- For nonlinear functions such as:

$$
f(a)=a^2
$$

$$\text{and:}$$

$$
f(a)=\ln(a)
$$

- The derivative depends on $a$, so the **slope changes at different points on the curve**.


* Key Takeaways

- **Derivative means slope.** It tells us how much the output changes in response to a tiny change in the input.

$$
\text{Derivative}=\text{Slope}
$$

- For a tiny change in $a$:

$$
\Delta f
\approx
\frac{df}{da}\Delta a
$$

- The derivative can be **different at different points**:

$$
f(a)=a^2
$$

$$
\text{At }a=2:\qquad \frac{df}{da}=4
$$

$$
\text{At }a=5:\qquad \frac{df}{da}=10
$$

- Standard derivative formulas can be looked up and used to determine the slope of common functions:

$$
\frac{d}{da}a^2=2a
$$

$$
\frac{d}{da}a^3=3a^2
$$

$$
\frac{d}{da}\ln(a)=\frac{1}{a}
$$

- The central intuition is:

$$
\boxed{
\text{Tiny change in output}
\approx
\text{Derivative}
\times
\text{Tiny change in input}
}
$$

### Computation Graphs

* A neural network's computation is organized as a **forward pass** (forward propagation) that computes the output, followed by a **backward pass** (backpropagation) that computes gradients/derivatives.
* The computation graph explains why it is organized this way. To illustrate it, use a simpler function than logistic regression:

$$
J(a,b,c)=3(a+bc)
$$

* Computing $J$ takes three distinct steps:

$$
u = bc
$$

$$
v = a + u
$$

$$
J = 3v
$$

* These steps can be drawn as a graph: the inputs $a,b,c$ on the left, each operation is a node, and the final output $J$ is on the right.

$$
(b,c)\rightarrow u=bc
\quad\rightarrow\quad
(a,u)\rightarrow v=a+u
\quad\rightarrow\quad
J=3v
$$

* Concrete example with $a=5$, $b=3$, $c=2$:

$$
u = bc = 3\cdot 2 = 6
$$

$$
v = a+u = 5+6 = 11
$$

$$
J = 3v = 3\cdot 11 = 33
$$

* Check by expanding directly: $3(5+3\cdot 2)=33$.
* The computation graph is useful when there is a **special output variable** we want to optimize. For logistic regression, $J$ is the cost function we want to minimize.
* A **left-to-right** pass (forward, blue arrows) computes the value of $J$.
* A **right-to-left** pass (backward, red arrows) computes the derivatives. This is the natural direction for derivative calculations.

* **Derivatives with a computation graph:** figure out how $J$ changes when we nudge each variable, working backward through the graph.

* **Derivative with respect to $v$ (one step back)**
* $J = 3v$ and currently $v=11$. If $v$ is bumped from $11$ to $11.001$, then $J$ goes from $33$ to $33.003$.
* $J$ increases 3 times as much as $v$, so:

$$
\frac{dJ}{dv}=3
$$

* This is the same idea as $f(a)=3a \Rightarrow \frac{df}{da}=3$, with $J$ playing the role of $f$ and $v$ the role of $a$.

* **Derivative with respect to $a$ (the chain rule)**
* If $a$ is bumped from $5$ to $5.001$, then $v=a+u$ goes from $11$ to $11.001$, and $J$ goes from $33$ to $33.003$. So $\frac{dJ}{da}=3$.
* Breaking this down: changing $a$ changes $v$, and changing $v$ changes $J$.
* This is the **chain rule**: if $a$ affects $v$ which affects $J$, then the change in $J$ per nudge of $a$ is the product of the two changes.

$$
\frac{dJ}{da}=\frac{dJ}{dv}\cdot\frac{dv}{da}
$$

* Here $\frac{dv}{da}=1$ because $v$ changes by the same amount as $a$:

$$
\frac{dJ}{da}=3\cdot 1=3
$$

* Computing $\frac{dJ}{dv}$ first helps compute $\frac{dJ}{da}$. This is another step of the backward calculation.

* **Notation used in code**
* In code there is usually a **final output variable** we care about (here $J$, often the loss $L$). Most computations are derivatives of this final output with respect to various intermediate quantities ($a,b,c,u,v$).
* Writing a long name like `dFinalOutputVar_dvar` is impractical, and since we always differentiate $J$, we use the shorthand `dvar`:

$$
\texttt{dvar} \;\equiv\; \frac{dJ}{d\,\texttt{var}}
$$

* So in code:

$$
\texttt{dv}=\frac{dJ}{dv}=3
$$

$$
\texttt{da}=\frac{dJ}{da}=3
$$

* **Derivative with respect to $u$**
* If $u$ is bumped from $6$ to $6.001$, then $v$ goes from $11$ to $11.001$ and $J$ goes from $33$ to $33.003$. This is the same analysis as for $a$.

$$
\frac{dJ}{du}=\frac{dJ}{dv}\cdot\frac{dv}{du}=3\cdot 1=3
$$

* So `du` $=3$.

* **Derivative with respect to $b$ and $c$**
* Using the chain rule, a change in $b$ first affects $u$, which then affects $J$:

$$
\frac{dJ}{db}=\frac{dJ}{du}\cdot\frac{du}{db}
$$

* Since $u=bc$ and $c=2$, bumping $b$ from $3$ to $3.001$ moves $u$ from $6$ to $6.002$. So $u$ increases twice as much as $b$:

$$
\frac{du}{db}=c=2
$$

* We already know $\frac{dJ}{du}=3$, so when $u$ goes up by $0.002$, $J$ goes up by $0.006$:

$$
\frac{dJ}{db}=3\cdot 2=6
$$

* Verify forward: $b=3.001 \Rightarrow u=6.002 \Rightarrow v=11.002 \Rightarrow J=33.006$. So `db` $=6$.
* Similarly for $c$, using $\frac{du}{dc}=b=3$:

$$
\frac{dJ}{dc}=\frac{dJ}{du}\cdot\frac{du}{dc}=3\cdot 3=9
$$

* So `dc` $=9$.

* **Summary of the backward pass**
$$
\frac{dJ}{dv}=3
\;\rightarrow\;
\frac{dJ}{da}=3,\;\frac{dJ}{du}=3
\;\rightarrow\;
\frac{dJ}{db}=6,\;\frac{dJ}{dc}=9
$$

* Key takeaways:
 * The forward (left-to-right) pass computes the output $J$, such as the cost function we want to optimize.
 * The backward (right-to-left) pass computes derivatives. It is the most efficient way because each derivative reuses the ones already computed closer to the output.
 * $\frac{dJ}{dv}$ is used to get $\frac{dJ}{da}$ and $\frac{dJ}{du}$, and $\frac{dJ}{du}$ is used to get $\frac{dJ}{db}$ and $\frac{dJ}{dc}$.
* This will next be applied to logistic regression to compute the derivatives needed for that model.
